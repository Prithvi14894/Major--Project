```python
import os
import shutil

# Correct source folder
source_folder = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\ODIR\ODIR-5K\ODIR-5K\Training Images"

# Destination folders
normal_folder = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Final_Dataset\normal"
dr_folder = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Final_Dataset\diabetic_retinopathy"
glaucoma_folder = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Final_Dataset\glaucoma"

normal_count = 0
dr_count = 0
glaucoma_count = 0

for _, row in df.iterrows():
    filename = row["filename"]
    source = os.path.join(source_folder, filename)

    if not os.path.exists(source):
        continue

    # Copy Normal
    if row["N"] == 1:
        shutil.copy2(source, os.path.join(normal_folder, filename))
        normal_count += 1

    # Copy Diabetic Retinopathy
    if row["D"] == 1:
        shutil.copy2(source, os.path.join(dr_folder, filename))
        dr_count += 1

    # Copy Glaucoma
    if row["G"] == 1:
        shutil.copy2(source, os.path.join(glaucoma_folder, filename))
        glaucoma_count += 1

print("✅ ODIR images copied successfully!")
print(f"Normal: {normal_count}")
print(f"Diabetic Retinopathy: {dr_count}")
print(f"Glaucoma: {glaucoma_count}")
```

```python
import os
import shutil

# APTOS dataset path
aptos_folder = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\APTOS"

# Final dataset folders
normal_folder = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Final_Dataset\normal"
dr_folder = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Final_Dataset\diabetic_retinopathy"

normal_count = 0
dr_count = 0

# Class 0 -> Normal
class0 = os.path.join(aptos_folder, "0")
for img in os.listdir(class0):
    src = os.path.join(class0, img)
    dst = os.path.join(normal_folder, img)
    shutil.copy2(src, dst)
    normal_count += 1

# Classes 1,2,3,4 -> Diabetic Retinopathy
for cls in ["1", "2", "3", "4"]:
    class_folder = os.path.join(aptos_folder, cls)

    for img in os.listdir(class_folder):
        src = os.path.join(class_folder, img)
        dst = os.path.join(dr_folder, img)
        shutil.copy2(src, dst)
        dr_count += 1

print("✅ APTOS merged successfully!")
print(f"Normal images added: {normal_count}")
print(f"DR images added: {dr_count}")
```

```python
import os
import shutil
import pandas as pd

# Load SMDG metadata
df = pd.read_csv(
    r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\metadata - standardized.csv"
)

# Source folder containing all SMDG fundus images
source_folder = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\full-fundus\full-fundus"

# Destination glaucoma folder
glaucoma_folder = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Final_Dataset\glaucoma"

copied = 0
missing = 0

for _, row in df.iterrows():

    # Copy only glaucoma images (types = 1)
    if row["types"] != 1:
        continue

    # Get image filename
    filename = os.path.basename(row["fundus"])

    source = os.path.join(source_folder, filename)
    destination = os.path.join(glaucoma_folder, filename)

    if os.path.exists(source):
        shutil.copy2(source, destination)
        copied += 1
    else:
        missing += 1

print("✅ SMDG Glaucoma images copied successfully!")
print(f"Glaucoma images copied: {copied}")
print(f"Missing images: {missing}")
```

```python
import os
from PIL import Image

dataset_path = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Final_Dataset"

corrupted = []

for category in os.listdir(dataset_path):
    category_path = os.path.join(dataset_path, category)

    for img in os.listdir(category_path):
        img_path = os.path.join(category_path, img)

        try:
            with Image.open(img_path) as image:
                image.verify()
        except Exception:
            corrupted.append(img_path)

print(f"Corrupted Images: {len(corrupted)}")

if corrupted:
    for img in corrupted:
        print(img)
```

```python
import os
from PIL import Image

# Original dataset
input_dataset = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Final_Dataset"

# Processed dataset
output_dataset = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Processed_Dataset"

# Create output folders
classes = ["normal", "diabetic_retinopathy", "glaucoma"]

for cls in classes:
    os.makedirs(os.path.join(output_dataset, cls), exist_ok=True)

target_size = (224, 224)

for cls in classes:

    input_folder = os.path.join(input_dataset, cls)
    output_folder = os.path.join(output_dataset, cls)

    count = 0

    for image_name in os.listdir(input_folder):

        input_path = os.path.join(input_folder, image_name)
        output_path = os.path.join(output_folder, image_name)

        try:
            img = Image.open(input_path)
            img = img.convert("RGB")
            img = img.resize(target_size)

            img.save(output_path)

            count += 1

        except Exception:
            print("Skipped:", image_name)

    print(f"{cls}: {count} images resized.")

print("\n✅ Dataset resized successfully!")
```

```python
import os
import shutil
from sklearn.model_selection import train_test_split

# Source folder (resized dataset)
source_dataset = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Processed_Dataset"

# Destination folder
split_dataset = r"D:\DR AND GLAUCOMA DETECTION PROJECT\datasets\Split_Dataset"

classes = ["normal", "diabetic_retinopathy", "glaucoma"]

# Create folder structure
for split in ["train", "validation", "test"]:
    for cls in classes:
        os.makedirs(os.path.join(split_dataset, split, cls), exist_ok=True)

for cls in classes:

    source_folder = os.path.join(source_dataset, cls)

    images = os.listdir(source_folder)

    # 80% train, 20% temp
    train_imgs, temp_imgs = train_test_split(
        images,
        test_size=0.20,
        random_state=42,
        shuffle=True
    )

    # 10% validation, 10% test
    val_imgs, test_imgs = train_test_split(
        temp_imgs,
        test_size=0.50,
        random_state=42,
        shuffle=True
    )

    # Copy training images
    for img in train_imgs:
        shutil.copy2(
            os.path.join(source_folder, img),
            os.path.join(split_dataset, "train", cls, img)
        )

    # Copy validation images
    for img in val_imgs:
        shutil.copy2(
            os.path.join(source_folder, img),
            os.path.join(split_dataset, "validation", cls, img)
        )

    # Copy test images
    for img in test_imgs:
        shutil.copy2(
            os.path.join(source_folder, img),
            os.path.join(split_dataset, "test", cls, img)
        )

    print(f"{cls} completed!")

print("\n✅ Dataset split successfully!")
```
